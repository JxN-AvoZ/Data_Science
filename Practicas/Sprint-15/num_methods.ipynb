El servicio de venta de autos usados Rusty Bargain está desarrollando una aplicación para atraer nuevos clientes. Gracias a esa app, puedes averiguar rápidamente el valor de mercado de tu coche. Tienes acceso al historial: especificaciones técnicas, versiones de equipamiento y precios. Tienes que crear un modelo que determine el valor de mercado.
A Rusty Bargain le interesa:
- la calidad de la predicción;
- la velocidad de la predicción;
- el tiempo requerido para el entrenamiento

## 1. Preparación de datos

###   1.1 Importación de librerías

```python
import pandas as pd
import numpy as np
import time
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import lightgbm as lgb

# 1. Cargar datos
df = pd.read_csv('/datasets/car_data.csv')

# Exploración inicial
print(df.shape)
print(df.info())
print(df.describe())
print(df.head())
```

### 1.2 Preprocesamiento

```python
# Eliminar columnas irrelevantes
df = df.drop(['DateCrawled', 'DateCreated', 'LastSeen', 
              'NumberOfPictures', 'PostalCode'], axis=1)

# Filtrar anomalías en Price
df = df[df['Price'].between(100, 20000)]

# Filtrar anomalías en RegistrationYear
df = df[df['RegistrationYear'].between(1950, 2016)]

# Filtrar anomalías en Power
df = df[df['Power'].between(10, 1000)]

# Rellenar nulos en columnas categóricas con 'unknown'
cat_cols = ['VehicleType', 'Gearbox', 'Model', 'FuelType', 'NotRepaired']
df[cat_cols] = df[cat_cols].fillna('unknown')

print(f"Filas tras limpieza: {len(df)}")
print(df.isnull().sum())
print(df.describe())
```

```python
# Resumen del preprocesamiento

print("=== Resumen del preprocesamiento ===")
print(f"Filas originales:        354,369")
print(f"Filas tras limpieza:     {len(df)}")
print(f"Filas eliminadas:        {354369 - len(df)}")
print(f"Columnas eliminadas:     DateCrawled, DateCreated, LastSeen, NumberOfPictures, PostalCode")
print(f"Nulos restantes:         {df.isnull().sum().sum()}")
```

## 2. Entrenamiento del modelo

### 2.1 Preparar features y target

```python
target = df['Price']
features = df.drop('Price', axis=1)

# Codificar variables categóricas con OrdinalEncoder
cat_cols = ['VehicleType', 'Gearbox', 'Model', 'FuelType', 'NotRepaired', 'Brand']
encoder = OrdinalEncoder()
features[cat_cols] = encoder.fit_transform(features[cat_cols])

# División train/test 80:20
X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.2, random_state=12345
)

print(f"Train: {X_train.shape}, Test: {X_test.shape}")
```

### 2.2 Entrenamiento y comparativa entre modelos

```python
# Modelo 1 — Regresión Lineal (prueba de cordura)

start = time.time()
lr = LinearRegression()
lr.fit(X_train, y_train)
train_time_lr = time.time() - start

start = time.time()
y_pred_lr = lr.predict(X_test)
pred_time_lr = time.time() - start

rmse_lr = mean_squared_error(y_test, y_pred_lr) ** 0.5
print(f"Regresión Lineal — RMSE: {rmse_lr:.2f} | Train: {train_time_lr:.2f}s | Pred: {pred_time_lr:.4f}s")
```

```python
# Modelo 2 — Árbol de Decisión

start = time.time()
dt = DecisionTreeRegressor(max_depth=8, random_state=12345)
dt.fit(X_train, y_train)
train_time_dt = time.time() - start

start = time.time()
y_pred_dt = dt.predict(X_test)
pred_time_dt = time.time() - start

rmse_dt = mean_squared_error(y_test, y_pred_dt) ** 0.5
print(f"Árbol de Decisión — RMSE: {rmse_dt:.2f} | Train: {train_time_dt:.2f}s | Pred: {pred_time_dt:.4f}s")
```

```python
# Modelo 3 — Random Forest

start = time.time()
rf = RandomForestRegressor(n_estimators=100, max_depth=10, random_state=12345, n_jobs=-1)
rf.fit(X_train, y_train)
train_time_rf = time.time() - start

start = time.time()
y_pred_rf = rf.predict(X_test)
pred_time_rf = time.time() - start

rmse_rf = mean_squared_error(y_test, y_pred_rf) ** 0.5
print(f"Random Forest — RMSE: {rmse_rf:.2f} | Train: {train_time_rf:.2f}s | Pred: {pred_time_rf:.4f}s")
```

```python
# Modelo 4 — LightGBM

start = time.time()
lgbm = lgb.LGBMRegressor(n_estimators=200, learning_rate=0.05, 
                          num_leaves=31, random_state=12345, n_jobs=-1)
lgbm.fit(X_train, y_train)
train_time_lgbm = time.time() - start

start = time.time()
y_pred_lgbm = lgbm.predict(X_test)
pred_time_lgbm = time.time() - start

rmse_lgbm = mean_squared_error(y_test, y_pred_lgbm) ** 0.5
print(f"LightGBM — RMSE: {rmse_lgbm:.2f} | Train: {train_time_lgbm:.2f}s | Pred: {pred_time_lgbm:.4f}s")
```

## 3. Análisis del modelo

```python
# Análisis de velocidad y calidad

results = pd.DataFrame({
    'Modelo': ['Regresión Lineal', 'Árbol de Decisión', 'Random Forest', 'LightGBM'],
    'RMSE': [rmse_lr, rmse_dt, rmse_rf, rmse_lgbm],
    'Tiempo entrenamiento (s)': [train_time_lr, train_time_dt, train_time_rf, train_time_lgbm],
    'Tiempo predicción (s)': [pred_time_lr, pred_time_dt, pred_time_rf, pred_time_lgbm]
}).sort_values('RMSE').reset_index(drop=True)

print("=== Comparación de modelos ===")
print(results.to_string(index=False))
```

```python
print("Mejor RMSE:              LightGBM (1659.05)")
print("Más rápido en training:  Regresión Lineal (0.04s)")
print("Mejor balance calidad/velocidad: LightGBM")
```

## 4. Conclusiones

1. La __regresión lineal__ $(RMSE=2919)$ confirmó que el pipeline es correcto
   ya que todos los modelos la superan con éxito.

2. __LightGBM__ es el modelo recomendado para Rusty Bargain:
   - Mejor RMSE: $1659$
   - Entrenamiento _4x_ más rápido que __Random Forest__ (3s vs 13s)
   - Predicciones rápidas para uso en producción

3. __Random Forest__ ofrece buena calidad ($RMSE=1845$) pero es el más
   lento en entrenamiento (13s), lo que lo hace menos atractivo.

4. El __árbol de decisión__ es el más rápido pero sacrifica calidad.
